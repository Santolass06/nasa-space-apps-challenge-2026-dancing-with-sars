# NISAR exploration — Dancing with the SARs

Scratch notebook to explore a study area before wiring it into the API/UI.

1. Search NISAR products for a bbox and date range (ASF DAAC).
2. Load a product, inspect polarisations and backscatter statistics.
3. Build a quick before/after comparison.

Requires an Earthdata login and `asf_search`, `rioxarray`, `matplotlib` installed in the
backend virtualenv. Data is written to `data/raw` and never committed.

```python
from pathlib import Path

import matplotlib.pyplot as plt

DATA = Path("../data")
(DATA / "raw").mkdir(parents=True, exist_ok=True)

# Study area: replace with the place your story is about (min_lon, min_lat, max_lon, max_lat)
BBOX = (-122.5, 37.7, -121.9, 38.1)
START = "2026-01-01"
END = "2026-03-01"
```

```python
import asf_search as asf

wkt = f"POLYGON(({BBOX[0]} {BBOX[1]}, {BBOX[2]} {BBOX[1]}, {BBOX[2]} {BBOX[3]}, {BBOX[0]} {BBOX[3]}, {BBOX[0]} {BBOX[1]}))"

results = asf.search(
    platform="NISAR",
    intersectsWith=wkt,
    start=START,
    end=END,
    maxResults=50,
)
print(f"{len(results)} products")
for product in results[:10]:
    print(product.properties.get("sceneName"), product.properties.get("startTime"))
```
